```python
import pandas as pd

df = pd.read_csv("D:\dvc\py\csv files\winequalityN.csv")
df
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
# remove the null values wali rows

df = df.dropna()
df.isna().sum()
df.info()
```

```python
x = df.iloc[:,0:13]
x = pd.get_dummies(x, drop_first=True)

y = df["quality"]
x
print("="*100)
print(y)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.linear_model import LinearRegression

x_train, x_test, y_train, y_test = train_test_split(x,y,train_size=0.87, random_state=56) # model is ovefitted here...
lr = LinearRegression()
lr.fit(x_train, y_train)

y_pred = lr.predict(x_test)

print("mae = ", mean_absolute_error(y_test, y_pred))
print("mse = ", mean_squared_error(y_test, y_pred))
print("r2 score = ", r2_score(y_test, y_pred))
print("coefficent = ", lr.coef_)
print("intercept point = ", lr.intercept_)
```
