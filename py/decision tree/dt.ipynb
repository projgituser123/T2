```python
# this is also classification based algorithmm..
# it forms a tree structure and a decision is observed at branches of the tree..
# it works 2 criteria : "information gain" or "gini impurity"
# entropy = is a measure of uncertainity or error or we can say : state of randomness
# je column ma information gain vadhare toh ee col ni dependecy vadhare hoi output calculate karwa mate 
# agar information gain is negligible then you can drop that col
# the final tree (kind of a flowchart) is made from level of dependency obtained at each level.
# jeno information gain vadhare hoy eeno node decision tree ma top to bottom decide thai..
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
from sklearn.metrics import accuracy_score, confusion_matrix

df = pd.read_csv("D:\dvc\py\csv files\DecisionTreeDataset.csv")
df
```

```python
x = df.iloc[:, 0:4]
x  = pd.get_dummies(x, drop_first=True)
y = df["Job Offered"]
# since the dataset is small we will directly make the decision tree, and will not split the data
dt = DecisionTreeClassifier(criterion="entropy", max_depth=2) # default criterion is "gini"
dt.fit(x,y)
print("get the depth of the tree = " ,dt.get_depth()) 
print(tree.export_text(dt)) # to get the structure of the tree from dataset or use this to print the tree.

y_pred = dt.predict(x)
cm = confusion_matrix(y, y_pred)
print("confusion matrix = \n", cm)
print("accuracy_score = ", accuracy_score(y,y_pred))
```
