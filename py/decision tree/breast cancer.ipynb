```python
import pandas as pd
from sklearn import tree
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
import matplotlib.pyplot as plot
from sklearn.tree import plot_tree

df = pd.read_csv("D:\dvc\py\csv files\data_breast_cancer.csv")
df
```

```python
df.info()
df.isna().sum()
```

```python
x =df.iloc[:, 2:32]
y = df["diagnosis"]

x_train, x_test, y_train, y_test  = train_test_split (x,y, train_size=0.85, random_state=50)
dt = DecisionTreeClassifier(criterion="entropy", max_depth=32)
dt.fit(x_train, y_train)
y_pred = dt.predict(x_test)

cm = confusion_matrix(y_test, y_pred)
print("depth of the tree = ", dt.get_depth())
print("confusion_matrix = \n",cm)
print("accuracy rate = ", accuracy_score(y_test, y_pred))
print("DecisionTreeClassifier = \n", tree.export_text(dt))
print("classification report = \n", classification_report(y_test, y_pred))
```

```python
plot.figure(figsize=(20,20))
plot_tree(dt, filled=True,)
```
