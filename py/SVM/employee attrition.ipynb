```python
import pandas as pd
import numpy as np
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split

df = pd.read_csv("D:\dvc\py\csv files\Employee_Attrition.csv")
df
```

```python
x = df.iloc[:, 1:10]
y = df["Attrition"]

x = pd.get_dummies(x, drop_first=True)

x_train, x_test, y_train, y_test = train_test_split(x,y,train_size=0.87, random_state=42)
print("x train dataset number = " , x_train.shape[0])
print("x test dataset number = " , x_test.shape[0])
print("y train dataset number = " , y_train.shape[0])
print("y test dataset number = " , y_test.shape[0])

acc_list = []
for c in range(1,11):
    svm = SVC(kernel="rbf", random_state=46, C=c)
    svm.fit(x_train,y_train)
    y_pred = svm.predict(x_test)
    acc = accuracy_score(y_test, y_pred)
    acc_list.append(acc)
    print(f"for C = {c}, Accuracy =  {acc}")

max_acc = max(acc_list)
best_c = (acc_list.index(max_acc)) +1 # value of C cannot be zero so we have to add +1 to increase the index .. 
print("best c  = ", best_c)

svm = SVC(kernel="rbf", random_state=46, C=best_c)
svm.fit(x_train,y_train)
y_pred = svm.predict(x_test)

print("confusion_matrix = \n", confusion_matrix(y_test, y_pred))
print("accuracy score = ",accuracy_score(y_test, y_pred))
print("classification_report = \n", classification_report(y_test, y_pred))
```
