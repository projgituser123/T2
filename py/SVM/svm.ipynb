```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
df = pd.read_csv("D:\dvc\py\csv files\diabetes.csv")
df.info()
df
```

```python
x = df.iloc[:, 0:8]
y = df["Outcome"]

x_train, x_test, y_train, y_test = train_test_split(x,y,train_size=0.9, random_state=42)
print("x train dataset number = " , x_train.shape[0])
print("x test dataset number = " , x_test.shape[0])
print("y train dataset number = " , y_train.shape[0])
print("y test dataset number = " , y_test.shape[0])

s = SVC(kernel="rbf", random_state=42, C=10)
s.fit(x_train, y_train)
y_pred = s.predict(x_test)
print("confusion_matrix = \n", confusion_matrix(y_test, y_pred))
print("accuracy score = ",accuracy_score(y_test, y_pred))
print("classification_report = ", classification_report(y_test, y_pred))
```
