```python
import pandas as pd

df= pd.read_csv("D:\dvc\py\csv files\car data.csv")
print(df)
print("="*50)
df.info()
```

```python
df["age"]= 2026 - df["Year"]
print(df)
df.info()
```

```python
x = df[["age","Kms_Driven","Present_Price","Fuel_Type","Seller_Type","Transmission","Owner"]] # double bracket means DataFrame()
y = df["Selling_Price"] # single square bracket means DataSeries()

#LR doesnt accept object type when we writ lr.fit().... so we have to change the datatype of those "object" using .grt_dummies() method...
x = pd.get_dummies(x, drop_first=True)
x.info()
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
x_train, x_test, y_train, y_test = train_test_split(x,y, train_size=0.85, random_state=46)
lr = LinearRegression()
lr.fit(x_train, y_train)

y_pred = lr.predict(x_test)
print("mae = ", mean_absolute_error(y_test, y_pred))
print("mse = ", mean_squared_error(y_test, y_pred))
print("r2 score = ", r2_score(y_test, y_pred))
```

```python
c = lr.coef_
print("coefficient = " ,c )
print("intercept point = ", lr.intercept_)
```
