```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({"temp":[0,20,40,60,80,100],"pressure":[0.0002,0.0012,0.0060,0.0300,0.0900,0.2700]})

print(df)
x = df.iloc[:,0:1]
y = df.iloc[:,-1]
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x,y)
y_pred = model.predict(x)
plt.scatter(df["temp"],y)
plt.plot(df["temp"],y_pred)

```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=8)
model_poly = poly.fit_transform(x)
model=LinearRegression()
model.fit(model_poly,y)
y_pred1 = model.predict(model_poly)
plt.scatter(df["temp"],y)
plt.plot(df["temp"],y_pred)
plt.plot(df["temp"],y_pred1)
```
