```python
import pandas as pd
import matplotlib.pyplot as plot
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error , r2_score
from sklearn.preprocessing import PolynomialFeatures

df = pd.read_csv("D:\dvc\py\csv files\Ice_cream selling data.csv")
df
df.info()
df.isnull().sum()
```

```python
x = df[["Temperature (°C)"]]
y = df["Ice Cream Sales (units)"]
 
lr = LinearRegression()
lr.fit(x,y)
y_pred = lr.predict(x)
plot.scatter(df["Temperature (°C)"],y)
plot.plot(df["Temperature (°C)"],y_pred, color = "black")

print("mse = ", mean_squared_error(y ,y_pred))
print("r2_score = ", r2_score(y, y_pred))
```

```python
poly = PolynomialFeatures(degree = 5)
pmx = poly.fit_transform(x)
lr.fit(pmx,y)
y_pred1 = lr.predict(pmx)
plot.scatter(df["Temperature (°C)"],y)
plot.plot(df["Temperature (°C)"],y_pred, color = "black")
plot.plot(df["Temperature (°C)"],y_pred1, color = "red")

print("poly mse = ",mean_squared_error(y, y_pred1))
print("ploy r2 score = ", r2_score(y, y_pred1))
```
