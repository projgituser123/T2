```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.ensemble import RandomForestClassifier

df = pd.read_csv("D:\dvc\py\csv files\Employee_Attrition.csv")
df.info()
print(df.isna().sum())
df.head()

# drop the col
df.drop(columns="Employee_ID", inplace=True)
df.head()
```

```python
x = df.iloc[:,0:9]
y = df["Attrition"]

x = pd.get_dummies(x, drop_first=True)

x_train, x_test, y_train, y_test = train_test_split(x,y, train_size=0.8, random_state=42)
print("x train dataset number = " , x_train.shape[0])
print("x test dataset number = " , x_test.shape[0])
print("y train dataset number = " , y_train.shape[0])
print("y test dataset number = " , y_test.shape[0])

knn = KNeighborsClassifier(n_neighbors=25)
knn.fit(x_train, y_train)
y_pred = knn.predict(x_test)

print("confusion matrix = \n", confusion_matrix(y_test, y_pred))
print("accuracy_score = ", accuracy_score(y_test, y_pred))
print("classifcation report = ", classification_report(y_test, y_pred))
```

```python
# if the question ask to optimise and get the highest accuracy score from the give "k" neighbour nodes = then put main LOC inside the for loop 
acc_score_list = []
for k in range(1,26):
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(x_train, y_train)
    y_pred = knn.predict(x_test)
    acc = accuracy_score(y_test, y_pred)
    acc_score_list.append(acc)
    print(f"k = {k} and Accuracy = {acc}")

max_acc = max(acc_score_list)
best_k = (acc_score_list.index(max_acc)) + 1 # +1 becasue the index of list start with 0 and we want k=2 which has the highest accuracy and it is sitting at index number 1 
print(best_k)
knn = KNeighborsClassifier(n_neighbors=best_k)
knn.fit(x_train,y_train)
y_pred = knn.predict(x_test)

print("confusion_matrix = \n", confusion_matrix(y_test, y_pred))
print("accuracy score = ",accuracy_score(y_test, y_pred))
print("classification_report = ", classification_report(y_test, y_pred))
```

```python
rf = RandomForestClassifier(n_estimators=150, max_depth=12, criterion="entropy", random_state=42)
rf.fit(x_train, y_train)
y_pred1 = rf.predict(x_test)
print("RF confusion_matrix = \n", confusion_matrix(y_test, y_pred1))
print("RF accuracy score = ",accuracy_score(y_test, y_pred1))
print("RF classification_report = ", classification_report(y_test, y_pred1))
```
