```python
# do a prediction using SVM and random forest algorithm from csv dataset.
# split dataset 80-20% and take random states  = 42
# for random forest ::: n_estimators = 50 to 100( find the best value and use it for training the model ); random states =42, criterion = entropy; 
# for SVM ::: random states = 42; C=10; find accuracy score and cm 

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier

df = pd.read_csv("D:\dvc\py\csv files\Hotel Reservations.csv")
df.info()
print(df.isna().sum())
df
```

```python
x = df.iloc[:,1:18]
x =pd.get_dummies(x, drop_first=True)
y = df["booking_status"]

x_train, x_test, y_train, y_test = train_test_split(x,y, random_state=40, train_size=0.8)

acc_list = []
for n in range(50,101):
    rf = RandomForestClassifier(criterion="entropy", random_state=42, n_estimators=n)
    rf.fit(x_train, y_train)
    y_pred = rf.predict(x_test)
    acc = accuracy_score(y_test, y_pred )
    acc_list.append(acc)
    print(f"for n_estimator = {n}, Accuracy =  {acc}")

max_acc = max(acc_list)
best_nestimator = (acc_list.index(max_acc))+ 1
print("best n_estimator is = ", best_nestimator)

rf = RandomForestClassifier(criterion="entropy", random_state=42, n_estimators= best_nestimator)
rf.fit(x_train, y_train)
y_pred = rf.predict(x_test)

print("accuracy_score = ", accuracy_score(y_test, y_pred))
print("cm = \n", confusion_matrix(y_test, y_pred))
print("classification report = ", classification_report(y_test, y_pred))

print("====" *80)
```

```python
svm = SVC(C=10, random_state=10, kernel="rbf")
svm.fit(x_train, y_train)
y_pred1 = svm.predict(x_test)

train_pred = svm.predict(x_train)
test_pred = svm.predict(x_test)

print("cm = \n", confusion_matrix(y_test, y_pred1))
print("classification report = ", classification_report(y_test, y_pred1))
train_acc = accuracy_score(y_train, train_pred)
test_acc = accuracy_score(y_test, test_pred)
print("train accuracy = ", train_acc)
print("test accuracy = ", test_acc)
```
